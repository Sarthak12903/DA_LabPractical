```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
commuter_times = np.array([16,8,35,17,13,15,15,5,16,25,20,20,12,10])
print("Sorted Array : " , np.sort(commuter_times))
```

```python
q1 = np.percentile(commuter_times,25)
q3 = np.percentile(commuter_times,75)
iqr = q3-q1
lower_bound = q1-1.5*iqr
upper_bound = q3 + 1.5*iqr
outlier = commuter_times[(commuter_times < lower_bound) | (commuter_times > upper_bound)]
print(outlier)
```

```python
print("5-number summary: " , commuter_times.min(),q1,np.median(commuter_times),q3,commuter_times.max())
print("Q1 = " ,q1)
print("Q3 = " , q3)
print("IQR = ", iqr )
print("lower_bound = ",lower_bound)
print("Upper bound = " , upper_bound)
print("Outlier Points = ", outlier)
```

```python
plt.Figure(figsize=(10,4))
sns.boxplot(x=commuter_times)
plt.title("Box Plot of Commuter times")
plt.xlabel("Minutes")
plt.show()
```
