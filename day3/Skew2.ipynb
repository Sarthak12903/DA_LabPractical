```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
airline_df = pd.read_csv("../Files/airline_passenger_satisfaction.csv")
```

```python
airline_df
```

```python
on_board_service = airline_df["On-board Service"]
```

```python
mean_rating = on_board_service.mean()
```

```python
median_rating = on_board_service.median()
```

```python
skewness_rating = on_board_service.skew()
```

```python
kurtosis_rating = on_board_service.kurt()
```

```python
print(on_board_service.value_counts().sort_index())
print(f"Mean on-board service: {mean_rating : .2f}")
```

```python
print(f"Median on-board service: {median_rating:.2f}")
print(f"Skewness of on-board service: {kurtosis_rating:.2f}")
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x=on_board_service,color="skyblue")
plt.axvline(mean_rating,color="red",linestyle="--",label=f"Mean : {mean_rating:.2f}")
plt.axvline(median_rating,color="black",linestyle="--",label=f"Median:{median_rating:.2f}")
plt.title(f"Distribution of on-board Service\n Skewness:{skewness_rating:.2f}, Kurtosis: {kurtosis_rating:.2f}")
plt.xlabel("on-board service rating")
plt.ylabel("Passengers")
plt.legend()
plt.tight_layout()
plt.show()
```
