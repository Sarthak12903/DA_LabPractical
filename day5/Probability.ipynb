```python
import numpy as np
import pandas as pd
import seaborn as sns 
```

```python
df = pd.read_csv("../Files/titanic.csv")
df
```

```python
df = df[['gender','survived']]
df = df.dropna()
```

```python
total = len(df)
print(f"Total passenger : {total}")

```

```python
total_female = len(df[df['gender'] == 'female'])
print(f"total female : {total_female}")
total_male = len(df[df['gender'] == 'male'])
print(f"Total male: {total_male}")

p_female = total_female/total
```

```python
p_female = total_female / total
p_male = total_male / total
```

```python
total_survived = len(df[df['survived'] == 1])
print(f"total survived: {total_survived}")
total_did_not_survive = len(df[df['survived'] == 0])
print(f"Total did not survive: {total_did_not_survive}")
```

```python
p_survived = total_survived / total
p_not_survived = total_did_not_survive / total
```

```python
print("\nMarginal Probilities:")
print(f"p(gender = female): {p_female: 3f}")
print(f"p(gender = male): {p_male: 3f}")
print(f"p(survived = 1): {p_survived: 3f}")
print(f"p(survived = 0): {p_not_survived: 3f}")
```

```python
p_female_survived = len(df[ (df['gender'] == 'female') &
(df['survived'] == 1)]) / total
p_female_not_survived = len(df[ (df['gender'] == 'female') &
(df['survived'] == 0)]) / total
p_male_not_survived = len(df[ (df['gender'] == 'male') &
(df['survived'] == 0)]) / total
```

```python
print("\nJoint Probabilities:")
print(f"P(female, survived=1): {p_female_survived:.3f}")
print(f"P(female, survived=0): {p_female_not_survived:.3f}")
print(f"P(male, survived=1):   {p_male_survived:.3f}")
print(f"P(male, survived=0):   {p_male_not_survived:.3f}")
```

```python
print("\nConditional Probabilities:")
p_survived_given_female = p_female_survived / p_female
print(f"P(survived=1 | gender=female):
{p_survived_given_female: .3f}")
```
