```python
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv("../Files/Amazon.csv",low_memory=False)
df
```

```python
cancelled_pending_orders = df[df["Status"].isin(['Cancelled','Pending'])]

contingency_table = pd.crosstab(cancelled_pending_orders["Status"],cancelled_pending_orders['Fulfilment'])
```

```python
print(contingency_table)
```

```python
chi2 , p , dof, ex = chi2_contingency(contingency_table)
print("Chi-Squared statistic : " , chi2)
print(f"P-Value : {p}")
print(f"Degree of freedom : {dof}")
print(f"Expected Values : {ex}")
```

```python
if p < 0.05:
    print(f"Reject null hypothesis")
else:
    print(f"Fail to reject null hypothesis")
```
