```python
sales = np.array([120,135,150,160,145,170,180,175,165,190,210,230])
months = np.array(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
```

```python
print("Total : " , sales.sum())
print("Mean : " , sales.mean())
print("Median : " , np.median(sales))
print("Min : " , sales.min(), "Max:", sales.max())
print("Range :" , np.ptp(sales))
print("var : " , sales.var().round(2))
print("Std : " , sales.std().round(2))
q1,q2,q3 = np.percentile(sales,[25,50,75])
print("Q1, Q2, Q3 : " , q1,q2,q3 , "IQR: ",q3-q1)
print("cumulative sales : " , np.cumsum(sales))
print("Best maonth : " , months[np.argmax(sales)],sales.max())
print("Worst Month : " , months[np.argmin(sales)],sales.min())
```

```python
expenses = np.array([2500,3200,np.nan,4100,3900,np.nan,3000])
print(expenses)

print(expenses.sum())
print(expenses.mean())
missing = np.isnan(expenses)
```

```python
print(missing)
```

```python
print("Missing count: " ,missing.sum())
```

```python
clean = expenses[~np.isnan(expenses)]
print(clean)
print(np.nan_to_num(expenses,nan=0))

filled = np.where(np.isnan(expenses),np.nanmean(expenses),expenses)
print(filled.round(2))
```
