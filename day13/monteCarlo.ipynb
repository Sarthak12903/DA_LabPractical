```python
import numpy as np
```

```python
def roll_dice():
    ## Simulation of rolling a dice twice , minimum number we get is 1 max is 6
    ## Add the results of the two simulations, so possible values will be : 
    ## (1,1) or (1,2) ....(6,6)....Sum will be between 2 and 12
    ## Run multiple times to verify 
    return np.sum(np.random.randint(1,7,2))
```

```python
print(roll_dice())
print(roll_dice())
print(roll_dice())
print(roll_dice())
print(roll_dice())
print(roll_dice())
```

### Part2 : Someone approaches us saying I will give you 5 dollars if you get 7 and take dollar if you get a number other than 7
### How do you we know that will happen?
### Our own "Monte Carlo Simulation " Like function



```python
def monte_carlo_simulation(runs=1000):
    result = np.zeros(2)
    for _ in range(runs):
        if roll_dice() == 7:
            result[0] += 1
        else:
            result[1] +=1
    return result
```

```python
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())
```

<p>Part 3 : Now o it 1000 times......Takes some time</p>

```python
results = np.zeros(1000)
for i in range(1000):
    results[i] = monte_carlo_simulation()[0]
print(results)
```

```python
import matplotlib.pyplot as plt
```

```python
fig,ax = plt.subplots()
ax.hist(results,bins=15)
plt.show()
```
