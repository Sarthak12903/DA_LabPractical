```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("../Files/titanic.csv")
df.head()
```

```python
selected_columns = ["survived","pclass","gender","age","sibsp","parch","embarked"]
df = df[selected_columns]
```

```python
x = df.drop("survived",axis=1)
y = df["survived"]
```

```python
categorical_features = ["pclass","gender","sibsp","parch","embarked"]
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
```

```python
encoded_data = encoder.fit_transform(df[categorical_features])
feature_name = encoder.get_feature_names_out(categorical_features)
```

```python

encoded_df = pd.DataFrame(encoded_data,columns=feature_name)
```

```python
missing_value_age = x["age"].isna().isna().any()
if missing_value_age:
    print("Handling nulls....")
    imputer = SimpleImputer(strategy="mean")
    imputer.fit(x["age"])
    x["age"] =imputer.transform(x[["age"]])[:,0]

print(x.age)
```

```python
scaler = StandardScaler()
scaler.fit(x[["age"]])
x_scaled = x.copy()
```

```python
print(x_scaled)
```

```python
x_scaled.loc[:,"age"] = scaler.transform(x[["age"]])
```

```python
print(x_scaled)
```

```python
df_merged = pd.concat([encoded_df,x_scaled["age"]],axis=1)
```
