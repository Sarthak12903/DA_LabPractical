```python
from scipy.stats import norm
```

```python
mu = 500
x_bar = 503.4
s = 10
n = 25
alpha  = 0.05
```

```python
se = s/n**0.5
```

```python
z = (x_bar - mu)/se
```

```python
critical_value = norm.ppf(1-alpha/2)
```

```python
p_value = 2*norm.sf(abs(z))
```

```python
print("Population mean = ", mu)
print("Sample Mean = ", x_bar)
print("Standard deviation = ", s)
print("Sample Size = ",n)
print("Alpha = ", alpha)
print("Standard error = ",se)
print("z Statistics = ",z)
print("critical_value = ", critical_value)
print("P-Value = ",p_value)

```

```python
if z > critical_value:
    print("critical_value Method : Rejected H0")
else:
    print("critical_value method : Do not reject H0")
```

```python
if p_value < alpha:
    print("P-Value methos : Rejected H0")
else: 
    print("P-value method: Do not reject H0")
```
