```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
```

```python
data = pd.read_csv("../Files/tips.csv")
data
```

```python
total_bill_Series = data["total_bill"]
```

```python
shapiro_test = stats.shapiro(total_bill_Series)
print(f"Test statistics (W) : {shapiro_test.statistic}")
print(f"p-value : {shapiro_test.pvalue}")
```

```python
alpha = 0.05
if shapiro_test.pvalue > alpha:
    print(f"Data is normally distributed")
else:
    print(f"Data is not normally distributed")
```

```python
## CLT

df = pd.read_csv("../Files/diabetes.csv")
df
```

```python
sample_means = []
n_samples = 100
sample_size = 30

for _ in range(n_samples):
    sample = df["Glucose"].sample(n=sample_size, replace=True)
    sample_means.append(sample.mean())
```

```python
plt.hist(sample_means,bins=30,edgecolor='k' , alpha=0.7)
plt.title("Distribution of sample Means (n=30)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```

```python
stats.probplot(sample_means,dist="norm",plot=plt)
plt.title("Q-Q Plot of Sample Means")
plt.show()
```

```python
shapiro_test = stats.shapiro(sample_means)
print(f"Shapiro_Wilk p_value : {shapiro_test.pvalue}")


if shapiro_test.pvalue < 0.05:
    print("Likely does not follow a normal distribution")
else:
    print(f"May be normally distributed.")
```
