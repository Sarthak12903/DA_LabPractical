```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import binom
```

```python
n = 5
k = 3
p = 0.66
```

```python
pmf = binom.pmf(k,n,p)
cdf = binom.cdf(k,n,p)
print(f"PMF (P( X = {k})) = {pmf:.4f}")
print(f"CDF (P(X <= {k})) = {cdf:.4f}")
```

```python
# PDF

prob = binom.cdf(3,n,p) - binom.cdf(1,n,p)
print(f"Binomial Probability P(2 <= X <= 3) = {prob:.4f}")
```

```python
# another methods for PDF# PDF

prob = binom.pmf(2,n,p) + binom.pmf(3,n,p)
print(f"Binomial Probability P(2 <= X <= 3) = {prob:.4f}")
```
