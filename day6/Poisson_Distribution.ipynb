```python
import numpy as np 
from scipy.stats import poisson
```

```python
x = 13
meu = 15

print(poisson.pmf(x,meu))
```

```python
x =8
meu = 10
print(poisson.pmf(x,meu))
```

```python
mu = 5
for x in range(5):
    print(poisson.pmf(x,mu))

print(poisson.cdf(2,mu))
print(1-poisson.cdf(3,mu))
```

```python
mu = 8
print(f"8 defective produced in day {poisson.pmf(8,mu)}")
```

```python
print(f"5 defective produced in day {poisson.pmf(5,mu)}")
```

```python
print(f"atmost 6 defective produced in day {poisson.cdf(6,mu)}")
```

```python
print(f"more than 10 defective produced in day {1- poisson.cdf(10,mu)}")
```

```python
print(f"between 6 and 9 defective produced in day { poisson.cdf(9,mu) - poisson.cdf(5,mu)}")
```

```python
## insurance
```

```python
mu = 3
print(f"P(sells atleast 1 policy) : {1 - poisson.pmf(0,mu)}")
```

```python
print(f"P(sells 2 to 4 policy) : {poisson.cdf(4,mu) - poisson.cdf(1,mu)}")
```

```python
print(f"P(sells exactly 5 policy) : { poisson.pmf(5,mu)}")
```

```python
print(f"P(sells atmost 1 policy) : { poisson.pmf(0,mu) + poisson.pmf(1,mu)}")
```
