```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Files/SP500 oil gold bitcoin.csv")

corr = df[["BITCOIN","Gold","S&P500"]].corr()
cov = df[["BITCOIN","Gold","S&P500"]].cov()

print(corr)
print(cov)

print(cov.to_string(float_format='{:.4f}'.format))

plt.figure(figsize=(10,6))
plt.scatter(df["BITCOIN"],df["Gold"], color='red',alpha=0.5,label='Bitcoin vs Gold')
plt.scatter(df["BITCOIN"],df["S&P500"],color='green',alpha=0.5,label="Bitcoin S&P")
plt.scatter(df["S&P500"],df["Gold"], color="yellow",alpha=0.5, label='S&P vs Gold')

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot=True, fmt=".2f", cmap="viridis")
plt.title("Overall Correlation Heatmap")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

# Use weight-height csv dataset . Find covariance , correlation betwwn height and weight for males , females and combines data. Plot all of them using scatter plot and heatmap


```python
df = pd.read_csv("../Files/weight-height.csv")
df
```

```python
male_details = df[df["Gender"] == "Male"]
female_details = df[df["Gender"] == "Female"]

print(male_details)
print(female_details)

corr_male = male_details[["Height", "Weight"]].corr()
cov_male = male_details[["Height","Weight"]].cov()

print(corr_male)
print(cov_male)

corr_female = female_details[["Height", "Weight"]].corr()
cov_female = female_details[["Height","Weight"]].cov()

print(corr_female)
print(cov_female)


combines_data_corr = df[["Height","Weight"]].corr()
combines_data_cov = df[["Height","Weight"]].cov()


print(combines_data_corr)
print(combines_data_cov)
```

```python
print(combines_data_cov.to_string(float_format='{:.4f}'.format))
print(cov_male.to_string(float_format='{:.4f}'.format))
print(cov_female.to_string(float_format='{:.4f}'.format))

plt.figure(figsize=(10,6))
plt.scatter(male_details["Height"],male_details["Weight"], color='red',alpha=0.5,label='Male_Height vs Male_Weight')
plt.scatter(female_details["Height"],female_details["Weight"],color='green',alpha=0.5,label="Female_Height vs Female_Weight")
plt.scatter(df["Height"],df["Weight"], color="pink",alpha=0.5, label='Combines_Height VS Combines_Weight')

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(combines_data_corr,annot=True, fmt=".2f", cmap="viridis")
plt.title("Overall Correlation Heatmap")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```
