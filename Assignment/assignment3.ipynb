### 14. Write a Python program to create a stacked bar plot with error bars. Note: Use bottom to stack the women?s bars on top of the men?s bars.

 - Means (women) = (25, 32, 30, 35, 29)

```python

groups = ["Group1", "Group2", "Group3", "Group4", "Group5"]
men_means = (22, 30, 35, 35, 26)
women_means = (25, 32, 30, 35, 29)
men_std = (4, 3, 4, 1, 5)
women_std = (3, 5, 2, 3, 3)

width = 0.35  


plt.bar(
    groups,
    men_means,
    width,
    yerr=men_std,
    color="red",
    edgecolor="black",
    ecolor="blue",
    capsize=3,
    label="Men",
)


plt.bar(
    groups,
    women_means,
    width,
    yerr=women_std,
    bottom=men_means,
    color="green",
    edgecolor="black",
    ecolor="green",
    capsize=3,
    label="Women",
)


plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Scores by group\nand gender")
plt.legend(loc="upper right")


plt.ylim(0, 80)

plt.show()
```

### 15. Write a Python program to create a horizontal bar chart with differently ordered colors. Note: Use bottom to stack the women?s bars on top of the men?s bars.
 - Sample Data Set:
 - languages = [['Language','Science','Math'],
 - ['Science','Math','Language'],
 - ['Math','Language','Science']]
 - numbers = [{'Language':75, 'Science':88, 'Math':96},
 - {'Language':71, 'Science':95, 'Math':92},
 - {'Language':75, 'Science':90, 'Math':89}]

```python

languages = [
    ["Language", "Science", "Math"],
    ["Science", "Math", "Language"],
    ["Math", "Language", "Science"],
]
numbers = [
    {"Language": 75, "Science": 88, "Math": 96},
    {"Language": 71, "Science": 95, "Math": 92},
    {"Language": 75, "Science": 90, "Math": 89},
]

students = ["Student-1", "Student-2", "Student-3"]
color_map = {"Language": "red", "Math": "green", "Science": "blue"}


y_pos = range(len(students))

for i, student in enumerate(students):
    left_offset = 0
    student_langs = languages[i]
    student_scores = numbers[i]

    for subj in student_langs:
        val = student_scores[subj]
        plt.barh(
            i,
            val,
            left=left_offset,
            color=color_map[subj],
            edgecolor="black",
            height=0.6,
        )
        left_offset += val


plt.yticks(y_pos, students)

plt.xlim(0, 300)

for subj, col in color_map.items():
    plt.barh(0, 0, color=col, label=subj)

plt.legend(loc="upper right")

plt.minorticks_on()
plt.grid(which="major", linestyle="-", linewidth=0.5, color="grey")
plt.grid(which="minor", linestyle=":", linewidth=0.5, color="lightgrey")

plt.show()
```

### 16. Write a Python program to create stack bar plot and add label to each section.
 - Sample data:
 - people = ('G1','G2','G3','G4','G5','G6','G7','G8')
 - segments = 4
#### multi-dimensional data
 - data = [[ 3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039,
 - 7.1123587, 12.77792868, 3.44773477],
 - [ 11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195,
 - 6.79685302, 7.24578743, 3.69371847],
 - [ 3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182,
 - 4.63832778, 11.16849999, 8.56883433],
 - [ 4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185,
 - 10.97567589, 3.98287652, 8.80552122]]
 - The code snippet gives the output shown in the following screenshot:

```python

people = ("G1", "G2", "G3", "G4", "G5", "G6", "G7", "G8")
segments = 4

data = np.array([
    [
        3.40022085,
        7.70632498,
        6.4097905,
        10.51648577,
        7.5330039,
        7.1123587,
        12.77792868,
        3.44773477,
    ],
    [
        11.24811149,
        5.03778215,
        6.65808464,
        12.32220677,
        7.45964195,
        6.79685302,
        7.24578743,
        3.69371847,
    ],
    [
        3.94253354,
        4.74763549,
        11.73529246,
        4.6465543,
        12.9952182,
        4.63832778,
        11.16849999,
        8.56883433,
    ],
    [
        4.24409799,
        12.71746612,
        11.3772169,
        9.00514257,
        10.47084185,
        10.97567589,
        3.98287652,
        8.80552122,
    ],
])


colors = ["red", "green", "white", "m"]

total_sum = np.sum(data)
percentages = (data / total_sum) * 100

y_pos = np.arange(len(people))
left_offsets = np.zeros(len(people))

for i in range(segments):
    widths = data[i]
    plt.barh(
        y_pos,
        widths,
        left=left_offsets,
        color=colors[i],
        edgecolor="black",
        height=0.75,
    )

    for j in range(len(people)):
        x_center = left_offsets[j] + widths[j] / 2.0
        pct_label = f"{int(round(percentages[i, j]))}%"
        plt.text(x_center, y_pos[j], pct_label, ha="center", va="center")

    left_offsets += widths
plt.yticks(y_pos, people)
plt.xlabel("Scores")
plt.xlim(0, 40)

plt.show()
```

### 17. Write a Python program to add textures (black and white) to bars and  wedges.
Note: Use bottom to stack the womens bars on top of the mens bars.

```python

patterns = ["|", "\\", "/", "+", "-", ".", "*", "x", "o", "O"]

for i, pattern in enumerate(patterns):
    plt.bar(
        i + 0.5,
        3,
        width=0.8,
        color="white",
        edgecolor="black",
        hatch=pattern,
        zorder=3,
    )


plt.xlim(0, 10)
plt.ylim(0, 3)

plt.show()
```
